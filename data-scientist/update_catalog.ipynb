```python
import json

#===================================
#Importo el archivo actual de catalog
#===================================

def load_catalog(filename="keyword_catalog.json"):
    """
    Load the keyword catalog from a JSON file.
    """
    with open(filename, "r", encoding="utf-8") as file:
        catalog = json.load(file)

    return catalog

#===================================
#Actualizo el catálogo con mi nueva lista de palabras
#===================================

def update_catalog(catalog, new_input):
  """
  Update the keyword catalog with new keywords.
  new_input must be a list

  """

  #uno la lista anterior con la nueva
  updated_catalog = catalog + new_input

  #elimino duplicados entre ambas listas
  updated_catalog = list(dict.fromkeys(updated_catalog))


  return updated_catalog

#===================================
#Genero un json actualizado del catálogo con el mismo nombre (lo debo subir al github para reemplazar el anterior)
#===================================

def save_catalog(updated_catalog, filename="keyword_catalog.json"):
    """
    Save the updated keyword catalog to a JSON file.
    """
    with open(filename, "w", encoding="utf-8") as file:
        json.dump(updated_catalog, file, ensure_ascii=False, indent=4)
```
