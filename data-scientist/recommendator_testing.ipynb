```python
import json
#!pip install oci

from recommendation_functions import (
    categories_for_recommendation,
    filter_by_category,
    calculate_keyword_matches,
    rank_by_matches,
    select_recommendations,
    format_recommendations
)
```

```python
with open("/content/knowledge_base.json", "r", encoding="utf-8") as file:
    knowledge_base = json.load(file)
```

```python
keywords = ["typescript","springboot"]
category = "Frontend"
confidence = 0.5
second_category = "Backend"
second_confidence = 0.4
```

```python
# Seleccionar las categorías a utilizar
categories = categories_for_recommendation(
    confidence,
    second_confidence,
    category,
    second_category
)

# Filtrar la base de conocimientos por categoría
filtered_items = filter_by_category(
    knowledge_base,
    categories
)

# Calcular coincidencias entre keywords y recursos
results_matches = calculate_keyword_matches(
    filtered_items,
    keywords
)

# Ordenar los recursos según el número de coincidencias
ranked_results = rank_by_matches(
    results_matches
)

# Seleccionar las recomendaciones finales
recommendations = select_recommendations(
    ranked_results,
    categories
)

# Formatear las recomendaciones para el resultado final
formatted_recommendations = format_recommendations(
    recommendations
)

formatted_recommendations
```

```python
output = {
    "category": category,
    "confidence": confidence,
    "keywords": keywords,
    "recommendations": formatted_recommendations
}

output
```
